## Goal
Ask ChatGPT to summarize a passage, critique its summary,
and revise it using clear requirements.

## Source Passage
A campus Python workshop teaches students basic programming
through short explanations and hands-on exercises. Students
practice variables, user input, loops, and working with data.
They can work in pairs and ask questions during the activities.
The workshop is designed for beginners, so no coding experience
is required. Students should bring a laptop. The session ends
with a small project that uses the skills they practiced.

## Initial Prompt
Summarize the passage above in one paragraph.

## Self-Reflection Prompt

Review your original summary against the source passage.

Check these requirements:
1. Accuracy: Use only facts from the passage.
2. Clarity: Use short sentences and simple words.
3. Audience: Write for students with no coding experience.
4. Length: Keep the revised summary under 60 words total.
5. Format: Use exactly three bullet points.
6. Coverage: Include the topics, hands-on practice, beginner
   eligibility, laptop requirement, and final project.

First, give a brief critique explaining what meets the requirements
and what needs improvement. Then provide the revised summary.
Do not invent dates, locations, or registration details.

## Self-Critique — ChatGPT Response

- Accuracy: The original summary matches the source passage.
- Clarity and audience: It is understandable, but the first
  sentence combines several details and can be shortened.
- Length: It is already under 60 words.
- Format: It is one paragraph instead of three bullet points.
- Coverage: The required details are included and should be
  preserved in the revision.

## Improved Summary — ChatGPT Response

- Learn Python variables, user input, loops, and working with data through short explanations and hands-on practice.
- No coding experience is needed. Bring a laptop.
- Work in pairs, ask questions, and finish with a small project using the skills you practiced.

## What Improved

The revision uses three bullet points and shorter sentences.
It is easier to scan and keeps the key details from the passage.

```python
# I saved the summaries from ChatGPT.
# This code checks formatting, not factual accuracy.

original_summary = (
    "The campus Python workshop introduces beginners to programming "
    "through short explanations and hands-on activities covering "
    "variables, user input, loops, and data. Students can work in pairs "
    "and ask questions, and the session ends with a small project. "
    "No coding experience is needed, but students should bring a laptop."
)

improved_bullets = [
    "Learn Python variables, user input, loops, and working with data "
    "through short explanations and hands-on practice.",

    "No coding experience is needed. Bring a laptop.",

    "Work in pairs, ask questions, and finish with a small project "
    "using the skills you practiced."
]

original_word_count = len(original_summary.split())
improved_word_count = sum(
    len(bullet.split()) for bullet in improved_bullets
)

print("ORIGINAL SUMMARY")
print(original_summary)

print("\nIMPROVED SUMMARY")
for bullet in improved_bullets:
    print("-", bullet)

print("\nCHECKS")
print("Original word count:", original_word_count)
print("Improved word count:", improved_word_count)
print("Bullet count:", len(improved_bullets))

assert improved_word_count < 60, "Summary must be under 60 words."
assert len(improved_bullets) == 3, "Summary must have three bullets."

print("Length and bullet count checks passed.")
```
