### Prompt
You are a Python coding assistant.

Create a function called calculate_total_sales(orders).

Requirements:
- Input: a list of dictionaries with "price" and "quantity".
- Calculate total sales by adding price times quantity for each order.
- Return a dictionary with total_sales and skipped_orders.
- An empty list should return zero sales and zero skipped orders.
- Skip orders with missing fields or invalid values.
- Price must be a finite, nonnegative number.
- Quantity must be a nonnegative integer.
- Reject booleans and numeric strings.
- Use only Python's standard library.
- Do not change the input.
- Display sales with two decimal places.

Follow these separate stages:

1. PLAN
Give a short implementation plan, including input validation.

2. ACT
Generate the function and runnable tests for valid orders,
an empty list, and orders with missing or invalid values.

3. OBSERVE
Wait for me to run the code and provide the actual outputs.
Do not claim you executed it.

4. FIX
Use my results to identify any problems, revise the code
if needed, and give me tests to run again.
If the tests pass, explain briefly what they confirm.

## Stage 1: Plan — ChatGPT Response

- Check that the input is a list.
- Validate each order's price and quantity.
- Skip invalid orders and count them.
- Calculate total sales from valid orders.
- Return total sales and the skipped count.
- Test valid orders, an empty list, and invalid orders.

## Stage 2: Act

I copied ChatGPT's generated code into the next cell and ran it.

```python
import math


def calculate_total_sales(orders):
    if not isinstance(orders, list):
        raise TypeError("orders must be a list")

    total_sales = 0.0
    skipped_orders = 0

    for order in orders:
        if not isinstance(order, dict):
            skipped_orders += 1
            continue

        price = order.get("price")
        quantity = order.get("quantity")

        valid_price = (
            type(price) in (int, float)
            and math.isfinite(price)
            and price >= 0
        )

        valid_quantity = (
            type(quantity) is int
            and quantity >= 0
        )

        if not valid_price or not valid_quantity:
            skipped_orders += 1
            continue

        total_sales += price * quantity

    return {
        "total_sales": total_sales,
        "skipped_orders": skipped_orders
    }


# Each test includes the expected total and skipped count.
test_cases = [
    (
        "Valid orders",
        [
            {"price": 10.0, "quantity": 2},
            {"price": 5.5, "quantity": 3}
        ],
        36.5,
        0
    ),
    ("Empty list", [], 0.0, 0),
    (
        "Mixed valid and invalid orders",
        [
            {"price": 10, "quantity": 2},
            {"price": 5},                         # Missing quantity
            {"price": -3, "quantity": 2},         # Negative price
            {"price": "8", "quantity": 1},        # Numeric string
            {"price": 4, "quantity": 1.5},        # Fractional quantity
            {"price": True, "quantity": 2},       # Boolean price
            {"price": 4, "quantity": False},      # Boolean quantity
            {"price": float("inf"), "quantity": 1},
            {"price": float("nan"), "quantity": 1},
            "invalid order",
            {"price": 0, "quantity": 3},          # Valid zero price
            {"price": 5, "quantity": 0}           # Valid zero quantity
        ],
        20.0,
        9
    )
]

for name, orders, expected_total, expected_skipped in test_cases:
    result = calculate_total_sales(orders)

    assert math.isclose(result["total_sales"], expected_total)
    assert result["skipped_orders"] == expected_skipped

    print(name)
    print(f"Total sales: ${result['total_sales']:.2f}")
    print("Skipped orders:", result["skipped_orders"])
    print("Test passed\n")
```

## Stage 4: Review — ChatGPT Response

The results match the expected values. The code calculates sales
correctly for the tested orders, handles an empty list, and skips
the tested invalid values.

No fixes were needed for these test cases. Two requirements still
need testing: rejecting an input that is not a list and keeping
the original input unchanged.

```python
import copy

# Check that the function does not change the input.
orders = [
    {"price": 10, "quantity": 2},
    {"price": "invalid", "quantity": 1}
]

original_orders = copy.deepcopy(orders)
calculate_total_sales(orders)

assert orders == original_orders
print("Input unchanged: Test passed")

# Check that an input that is not a list is rejected.
try:
    calculate_total_sales("invalid input")
except TypeError as error:
    print("Invalid input rejected:", error)
    print("Test passed")
else:
    raise AssertionError("Expected a TypeError")
```
