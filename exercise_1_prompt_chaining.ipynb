## Goal
Build a customer support prompt chain where each step uses the
previous step's output.

## Customer Message
"My order says delivered, but I haven't received it."

## Chain
1. Classify the issue.
2. Identify missing information.
3. Suggest a solution.
4. Decide whether to escalate.

## Step 1: Classify the Issue

I used this prompt in ChatGPT to identify the customer's issue.

### Prompt
You are a customer support assistant.

Classify the customer message into one category:
delivery, billing, product, or other.

Return a JSON object with:
- category
- issue
- brief explanation

Use only information from the message.
Do not suggest a solution yet.

Customer message:
"My order says delivered, but I haven't received it."

### ChatGPT Response
{
  "category": "delivery",
  "issue": "Package marked delivered but not received",
  "brief explanation": "The customer has not received an order that is marked delivered."
}

```python
# I saved ChatGPT's response so I can use it in the next step.
# This stores the response. It does not make a live AI call.

customer_message = "My order says delivered, but I haven't received it."

step_1_output = {
    "category": "delivery",
    "issue": "Package marked delivered but not received",
    "brief explanation": (
        "The customer has not received an order that is marked delivered."
    )
}

print("Category:", step_1_output["category"])
print("Issue:", step_1_output["issue"])
print("Explanation:", step_1_output["brief explanation"])
```

## Step 2: Identify Missing Information

I used the classification from Step 1 to ask for the information
needed before suggesting a solution.

### Prompt
You are a customer support assistant.

Use the previous step's classification to identify missing information.

Ask no more than three short questions.
Use a polite, simple tone.
Do not ask for passwords or payment information.
Do not promise a refund or replacement.

Previous output:
{
  "category": "delivery",
  "issue": "Package marked delivered but not received",
  "brief explanation": "The customer has not received an order that is marked delivered."
}

Return a JSON object with a "questions" list.

### ChatGPT Response
{
  "questions": [
    "What is your order number?",
    "When was the package marked delivered?",
    "Have you checked nearby delivery spots and with neighbors?"
  ]
}

```python
import json

# I used Step 1's output to build the next prompt.
step_2_prompt = f"""
You are a customer support assistant.

Use the previous step's classification to identify missing information.

Ask no more than three short questions.
Use a polite, simple tone.
Do not ask for passwords or payment information.
Do not promise a refund or replacement.

Previous output:
{json.dumps(step_1_output, indent=2)}

Return a JSON object with a "questions" list.
"""

# Captured ChatGPT response, not a live API call.
step_2_output = {
    "questions": [
        "What is your order number?",
        "When was the package marked delivered?",
        "Have you checked nearby delivery spots and with neighbors?"
    ]
}

print("Prompt sent to ChatGPT:")
print(step_2_prompt)

print("Questions:")
for question in step_2_output["questions"]:
    print("-", question)
```

## Step 3: Suggest a Solution

I added a sample customer reply to answer the questions from Step 2.
This reply is fictional and is only used for testing.

### Sample Customer Reply
"My order number is ORD-1042. It was marked delivered two days ago.
I checked nearby delivery spots and asked my neighbors, but nobody has it."

### Prompt
You are a customer support assistant.

Use the issue classification, questions, and customer answers below
to suggest the next action.

Write a polite response under 80 words.
Acknowledge what the customer already checked.
Do not repeat questions they already answered.
Do not invent tracking details or company policies.
Do not promise a refund or replacement.
Suggest a support review if the package is still missing.

Issue:
Package marked delivered but not received.

Previous questions and customer answers:
- Order number: ORD-1042
- Marked delivered: two days ago
- Nearby delivery spots and neighbors checked: yes, package not found

Return JSON with "customer_response" and "next_action".

### ChatGPT Response
{
  "customer_response": "I'm sorry your package is still missing. Since you already checked nearby delivery spots and with your neighbors, the next step is to contact support with order ORD-1042 and request a delivery review.",
  "next_action": "Request a delivery review from support."
}

```python
# Fictional answers to the questions from Step 2.
customer_answers = [
    "ORD-1042",
    "Two days ago",
    "Yes. I checked nearby delivery spots and with neighbors. It was not found."
]

# I paired each question with its answer.
question_answers = [
    {"question": question, "answer": answer}
    for question, answer in zip(
        step_2_output["questions"], customer_answers
    )
]

step_3_prompt = f"""
You are a customer support assistant.

Use the issue classification, questions, and customer answers below
to suggest the next action.

Write a polite response under 80 words.
Acknowledge what the customer already checked.
Do not repeat questions they already answered.
Do not invent tracking details or company policies.
Do not promise a refund or replacement.
Suggest a support review if the package is still missing.

Issue classification:
{json.dumps(step_1_output, indent=2)}

Questions and customer answers:
{json.dumps(question_answers, indent=2)}

Return JSON with "customer_response" and "next_action".
"""

# Captured ChatGPT response, not a live API call.
step_3_output = {
    "customer_response": (
        "I'm sorry your package is still missing. "
        "Since you already checked nearby delivery spots and with "
        "your neighbors, the next step is to contact support with "
        "order ORD-1042 and request a delivery review."
    ),
    "next_action": "Request a delivery review from support."
}

print("Step 3 prompt:")
print(step_3_prompt)

print("Customer response:")
print(step_3_output["customer_response"])

print("\nNext action:")
print(step_3_output["next_action"])
```

## Step 4: Decide Whether to Escalate

I used Step 3's proposed action and the customer answers to decide
whether a human support agent should review the issue.

### Prompt
You are a customer support assistant.

Review the previous step's proposed action and customer answers.

For this exercise, use this test rule:
Escalate to a human support agent if the package remains missing
after the customer checked nearby delivery spots and with neighbors.
Otherwise, suggest those checks first.

This is a fictional test rule, not an actual company policy.

Return JSON with:
- escalate: true or false
- reason: a short explanation
- handoff_note: a short factual note for the support agent

Do not claim that you contacted support or opened a ticket.
Do not promise a refund or replacement.

Previous proposed action:
Request a delivery review from support.

Customer answers:
Order ORD-1042 was marked delivered two days ago.
The customer checked nearby delivery spots and with neighbors.
The package was not found.

### ChatGPT Response
{
  "escalate": true,
  "reason": "The package is still missing after the customer completed the suggested checks.",
  "handoff_note": "Please review order ORD-1042. The customer reports it was marked delivered two days ago but was not received. Nearby delivery spots and neighbors were checked without finding the package."
}

```python
step_4_prompt = f"""
You are a customer support assistant.

Review the previous step's proposed action and customer answers.

For this exercise, use this test rule:
Escalate to a human support agent if the package remains missing
after the customer checked nearby delivery spots and with neighbors.
Otherwise, suggest those checks first.

This is a fictional test rule, not an actual company policy.

Return JSON with:
- escalate: true or false
- reason: a short explanation
- handoff_note: a short factual note for the support agent

Do not claim that you contacted support or opened a ticket.
Do not promise a refund or replacement.

Previous step's output:
{json.dumps(step_3_output, indent=2)}

Customer questions and answers:
{json.dumps(question_answers, indent=2)}
"""

# Captured ChatGPT response, not a live API call.
step_4_output = {
    "escalate": True,
    "reason": (
        "The package is still missing after the customer "
        "completed the suggested checks."
    ),
    "handoff_note": (
        "Please review order ORD-1042. The customer reports it was "
        "marked delivered two days ago but was not received. "
        "Nearby delivery spots and neighbors were checked "
        "without finding the package."
    )
}

print("Step 4 prompt:")
print(step_4_prompt)

print("Escalate:", step_4_output["escalate"])
print("Reason:", step_4_output["reason"])
print("Handoff note:", step_4_output["handoff_note"])
```

## How I Ran the Chain
I used ChatGPT in this conversation to get the responses for each step.
I saved those responses in Python dictionaries and used them to build
the next prompts. This notebook demonstrates a manual prompt chain,
not an automated API workflow.
