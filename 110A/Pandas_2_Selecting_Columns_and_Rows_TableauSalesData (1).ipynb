## Getting columns from a `DataFrame`

The reference notebook explains that you can get columns out of a DataFrame in much the same way that you get elements out of a dictionary.

Reference: [Pandas DataFrame cookbook](http://pandas.pydata.org/pandas-docs/stable/cookbook.html#dataframes)


```python
# Method 1: bracket notation
# This method works for any column name.
sales["Region"].head()
```

### Method 2: dot notation

If a column name has **no spaces or special characters**, you can sometimes use dot notation.


```python
sales.Profit.head()
```

Dot notation does **not** work well for column names such as `Sub-Category` or `Customer Name`, so bracket notation is generally safer.


## A single column is a `Series`

When you select one column using one pair of brackets, Pandas returns a **Series**.


```python
print(type(sales["Profit"]))
sales["Profit"].head()
```

## Selecting multiple columns

To select multiple columns, provide a **list of column names** inside `[]`. Because the list itself also uses square brackets, the expression contains two sets of brackets: `[[]]`.


```python
sales[["Category", "Sub-Category", "Sales", "Profit"]].head(10)
```

```python
# When we select multiple columns, we get another DataFrame
business_df = sales[["Category", "Sub-Category", "Sales", "Profit"]]
print(type(business_df))
```

### Method 3: selecting columns with `.loc[]`

`.loc[]` can select rows and columns at the same time. A colon `:` means **all rows**.


```python
# all rows, but only these three columns
sales.loc[:, ["Customer Name", "Segment", "Region"]].head()
```

## Selecting a subset of rows

To select rows, we use the following approach:

1. generate a Boolean Series containing `True` or `False` for each row
2. use that Series to keep only the rows where the condition is `True`

Documentation: [Boolean indexing](https://pandas.pydata.org/docs/user_guide/indexing.html#boolean-indexing)

```python
# Create the condition - it is a Series with True/False values
is_west = (sales["Region"] == "West")
print(type(is_west))
is_west.head(10)
```

```python
# Use the condition to select only rows from the West
sales_west = sales[is_west]
sales_west.head(5)
```

```python
# Now we can ask questions about only the West
sales_west["Sub-Category"].value_counts().head(10)
```

You do not have to create a separate variable for the condition.

```python
# same idea as: SELECT * FROM sales WHERE Segment='Corporate'
corporate_sales = sales[sales["Segment"] == "Corporate"]
corporate_sales.head()
```

## Selecting rows with `.loc[]`

`.loc[]` is another important way to select rows by a condition. It becomes especially helpful when combining row filtering and column selection.

Documentation: [https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html)

```python
# keep only rows where the Sub-Category is Tables
tables = sales.loc[sales["Sub-Category"] == "Tables"]
tables.head()
```

Notice the difference between `=` and `==`:

- `=` assigns a value to a variable
- `==` compares two values

```python
# select rows AND only the columns we need
west_profit = sales.loc[sales["Region"] == "West", ["Region", "Sales", "Profit"]]
west_profit.head()
```

## Combining multiple conditions

Use:

- `&` for **AND**
- `|` for **OR**

Put each condition inside parentheses.

```python
# Tables sold to Corporate customers
corporate_tables = sales.loc[
    (sales["Sub-Category"] == "Tables") &
    (sales["Segment"] == "Corporate")
]
corporate_tables.head()
```

```python
# Transactions in either the West OR South region
west_or_south = sales.loc[
    (sales["Region"] == "West") |
    (sales["Region"] == "South")
]
west_or_south.head()
```

## Filtering numeric values

Conditions are not limited to text. We can also use comparisons such as `<`, `>`, `<=`, and `>=`.

```python
# find transactions that lost money
loss_sales = sales.loc[sales["Profit"] < 0]
loss_sales[["Sub-Category", "Sales", "Discount", "Profit"]].head(10)
```
