## Data Types in `DataFrame` columns

Pandas tries to determine the data type of each column based on the values it contains.

```python
# check the data types for each column
sales.dtypes
```

Common types you will see include:

- `object` — usually text/string data
- `int64` — whole numbers
- `float64` — numbers that may contain decimals
- `datetime64[ns]` — dates and times

Notice that `Postal Code` appears numeric even though conceptually a ZIP code is an identifier, not a quantity we would normally average.

## Descriptive statistics

The reference Pandas notebook introduces two especially useful methods:

- `.describe()` — get a quick overview of the data
- `.value_counts()` — find common values that appear in a column

Documentation: [Pandas `DataFrame.describe()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html)


```python
# for numeric columns
sales[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

For numeric columns, `.describe()` reports values such as **count, mean, standard deviation, minimum, quartiles, and maximum**.


```python
# for datetime columns
sales[["Order Date", "Ship Date"]].describe()
```

```python
# for text/categorical columns
sales[["Category", "Sub-Category", "Segment", "Region"]].describe()
```

### Frequency counts with `.value_counts()`

`value_counts()` tells us how often each distinct value appears.

Documentation: [Pandas `Series.value_counts()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.value_counts.html)


```python
# calculate frequency of the values
sales["Region"].value_counts()
```

```python
# there are more sub-categories, so let's view them in descending frequency
sales["Sub-Category"].value_counts()
```

```python
# let's get only the top 5
sales["Sub-Category"].value_counts().head(5)
```

### Exercise 3 — Descriptive statistics

```python
# Your code is here


print(df[["Sales", "Profit"]].describe())

print(df[["Category", "Segment"]].describe())

print(df["State"].value_counts().head(10))

print(df["Sub-Category"].value_counts().head(5))
```
