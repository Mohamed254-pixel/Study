## 1. Creating a list

```python
# A list is a series of items in a particular order.
# Square brackets indicate a list.
# Commas separate individual elements.
regions = ["south", "east", "west", "central"]
print(regions)
```

## 2. List indices

```python
# Python indices start at 0.
print("First item:", regions[0])
print("Second item:", regions[1])
print("Last item using -1:", regions[-1])
print("Length of list:", len(regions))

# Intentional IndexError example:
try:
    print(regions[4])
except IndexError as error:
    print("IndexError example:", error)
```

## 3. Modifying a list

```python
regions = ["south", "east", "west", "central"]
print("Original:", regions)

# Change a value by assigning a new value at an index.
regions[0] = "north"
print("After changing index 0:", regions)

# Add a value to the end with append().
regions.append("south")
print("After append:", regions)
```

## 4. Inserting and deleting values

```python
regions = ["south", "east", "west", "central"]
print("Original:", regions)

# Insert adds a new item at a specific position.
regions.insert(0, "north")
print("After insert:", regions)

# del removes an item by index.
del regions[0]
print("After del regions[0]:", regions)
```

## 5. Organizing a list

```python
regions = ["south", "east", "west", "central"]
print("Original:", regions)

# sort() permanently changes list order into ascending order (alphabetical for strings, numeric for numbers)
regions.sort()
print("After permanent sort:", regions)

regions = ["south", "east", "west", "central"]
print("Temporary sorted output:", sorted(regions))
print("Original list after sorted():", regions)

# reverse() permanently reverses the current list order.
regions.reverse()
print("After reverse:", regions)
```

## 6. pop() and remove()

```python
regions = ["south", "east", "west", "central"]
print("Original:", regions)

# pop() removes the last item by default and lets you use it.
popped_region = regions.pop()
print("After pop:", regions)
print("Popped item:", popped_region)

# pop(index) removes a specific index.
popped_region = regions.pop(1)
print("After pop(1):", regions)
print("Popped item:", popped_region)

# remove(value) removes the first matching value.
regions = ["south", "east", "west", "central", "south"]
regions.remove("south")
print("After remove('south'):", regions)
```

## 7. Lists are mutable

```python
# Mutable means the values can change.
regions = ["south", "east", "west", "central"]
regions[2] = "northwest"
print(regions)
```

## 8. Tuples

```python
# Tuples are like lists, but immutable.
# They use parentheses instead of square brackets.
dimensions = (200, 50)
print(dimensions[0])
print(dimensions[1])

# Intentional TypeError example:
try:
    dimensions[0] = 300
except TypeError as error:
    print("TypeError example:", error)

# You can overwrite a whole tuple.
dimensions = (300, 50)
print("New tuple:", dimensions)

# Or copy a tuple into a list to make the values mutable.
dimensions_list = list(dimensions)
dimensions_list[0] = 400
print("Mutable copy:", dimensions_list)
```

## 9. Mini practice: Store Inventory Manager

```python
# 1. Initialize your starting inventory list
inventory = ["laptops", "monitors", "keyboards", "chargers"]

print("📦 Initial Inventory:", inventory)
print("  ➜ First item in warehouse :", inventory[0])
print("  ➜ Most recently added item :", inventory[-1])

# 2. Add a newly arrived shipment to the end
new_item = "headphones"
inventory.append(new_item)
print(f"\n🚚 Arrived shipment: '{new_item}' added via .append()")

# 3. Alphabetize the stock for the store catalog
inventory.sort()
# 4. Display the final updated list and count
print("\n📋 Final Sorted Inventory Catalog:", inventory)
print("  ➜ Total distinct product categories:", len(inventory))
```
