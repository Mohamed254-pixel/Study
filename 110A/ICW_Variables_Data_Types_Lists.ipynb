## Part 2: Lists

### 1. Guest List
Create a list containing Yong-Mi, Jeff, Viet, Rui, and Samantha. Then use your list to print a message to each person, inviting them to dinner. The invite message can be simple, such as “Yong-Mi, please join us for a BBQ next Saturday at 6pm.”

```python
#Solution:
Guests = ["Yong-Mi", "Jeff", "Viet", "Rui", "Samantha"]
print(Guests)

for guest in Guests:
    print(guest + " please join us for a BBQ next Saturday at 6pm. ")

```

### 2. More Guests
You just found a bigger dinner table, so now more space is available. Invite Jason, Shawna, and Miguel.
* Use insert() to add Jason to the beginning of your list.
* Use insert() to add Shawna to the middle of your list.
* Use append() to add Miguel to the end of your list.
* Print a new set of invitation messages, one for each person in your list.


```python
#Solution:
Guests.insert(0, "Jason")
Guests.insert(3, "Shawna")
Guests.append("Miguel")

for guest in Guests:
    print(guest + " please join us for a BBQ next Saturday at 6pm. ")
```
