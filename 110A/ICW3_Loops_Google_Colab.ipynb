## 1. While Loop
Write a `while` loop that sums all the numbers from **1 to 5**.


````python
# ```python
# initial_condition
# while condition_is_true:
#     statement_to_repeat
#     update_condition
# ```
# Write your answer here


total = 0
number = 1

while number <= 5:
    total = total + number
    number = number + 1

print("Total:", total)
````

## 2. For Loop
Write a `for` loop that sums all the numbers from **1 to 5**.


````python
# ```python
# for x in range_or_list:
#     statement_to_repeat
# ```
# Write your answer here


total = 0

for number in range(1, 6):
    total = total + number

print("Total:", total)
````

## 3. Sum from 1 to n
Write a loop that sums all the numbers from **1 to n**, where `n` is a value given by the user.


````python
# Get n from the user
n = int(input("Enter n: "))

# While loop

while_total = 0
number = 1

while number <= n:
    while_total = while_total + number
    number = number + 1

print("While loop total:", while_total)

# For loop

for_total = 0

for number in range(1, n + 1):
    for_total = for_total + number

print("For loop total:", for_total)

````
